```python
#POST - Recipe into recipelist - local cli input
import requests

url = "http://127.0.0.1:8000/"

r_id = int(input("Welche ID kriegt das Rezept?"))
r_name = str(input("Wie heißt das Rezept?"))
r_dauer = int(input("Wie lange dauert das Rezept?"))
r_zutaten = input("Welche Zutaten hat das Rezept? (Zutaten durch ein Komma trennen)").split(", ")
r_schwierigkeit = str(input("Wie schwer ist das Rezept?"))
r_zubereitung = str(input("Wie bereitet man das Rezept zu?"))
r_bemerkungen = str(input("Hast du Bemerkungen zu dem Rezept?"))

payload_input = [
{
    "rezept_id": r_id,
    "name": r_name,
    "dauer": r_dauer,
    "zutaten": r_zutaten,
    "schwierigkeit": r_schwierigkeit,
    "zubereitung": r_zubereitung,
    "bemerkungen": r_bemerkungen
}
]

add_recipes_mask = requests.post(url+"rezepte", json=payload_input)

print(add_recipes_mask.json())
```

```python
#POST - Ingredientsearch - local cli input
import requests

url = "http://127.0.0.1:8000/"

r_zutaten_ingsearch = input("Mit welcher Zutat möchtest du kochen?")

payload_ingsearch = {
    "zutaten": r_zutaten_ingsearch
}

cli_ingredient_search = requests.post(url+"rezepte/zutaten", json=payload_ingsearch)

print(cli_ingredient_search.json())
```

```python
#PUT - Add comments to existing recipe and update data - local cli input
import requests

url = "http://127.0.0.1:8000/"

r_sel_id = str(input("Welche ID hat das Rezept zu dem du eine Bemerkung hinzufügen möchtest?"))
r_comment = str(input("Welche Bemerkung möchtest du hinzufügen?"))

payload_comments = {
    "bemerkungen": r_comment
}

add_comment = requests.put(url+"/rezepte/"+r_sel_id, json=payload_comments)

print(add_comment.json())
```
